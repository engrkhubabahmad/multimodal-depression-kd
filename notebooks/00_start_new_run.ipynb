# 00 - Start a new exact-v3 reproduction run

Creates a **brand-new Drive subfolder**. It never deletes or overwrites historical experiment folders. Run this once before notebooks 01–06.

```python
from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
from datetime import datetime, timezone
import json, platform, sys

DAIC = Path('/content/drive/MyDrive/DAIC_WOZ')
assert DAIC.is_dir(), f'Missing {DAIC}'

EXP_ROOT = DAIC / 'experiments' / 'exact_v3_reproduction'
EXP_ROOT.mkdir(parents=True, exist_ok=True)

run_id = 'run_' + datetime.now(timezone.utc).strftime('%Y%m%d_%H%M%S_UTC')
RUN = EXP_ROOT / run_id
RUN.mkdir(parents=False, exist_ok=False)

manifest = {
    'run_id': run_id,
    'purpose': 'exact frozen-v3 TRAIN107 DEV34 Standard-KD reproduction',
    'core_commit': 'd6ed3a8b54ec9cfa7c706cb441671c6e8b07096d',
    'train_participants': 107,
    'dev_participants': 34,
    'excluded_dev_participant': 440,
    'test_policy': 'NEVER OPEN OR SCORE TEST',
    'python': sys.version,
    'platform': platform.platform(),
}
(RUN/'run_manifest.json').write_text(json.dumps(manifest,indent=2)+'\n')
(EXP_ROOT/'ACTIVE_RUN.txt').write_text(run_id+'\n')

print('NEW RUN CREATED:', RUN)
print(json.dumps(manifest,indent=2))
```
