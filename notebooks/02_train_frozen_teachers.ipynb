This notebook is only a Colab runner. The frozen audio/text teacher implementation is stored in Git at `scripts/teachers/train_frozen_teachers.py`. Experiment outputs and feature caches stay in Google Drive.


## 0. Clone or update main


```python
from pathlib import Path
import base64, subprocess, sys
from google.colab import userdata

REPO='engrkhubabahmad/reliability-aware-depression-kd'
ROOT=Path('/content/reliability-aware-depression-kd')
token=userdata.get('GITHUB_TOKEN'); assert token, 'Add GITHUB_TOKEN in Colab Secrets.'
auth=base64.b64encode(f'x-access-token:{token}'.encode()).decode()
git=['git','-c',f'http.extraHeader=AUTHORIZATION: basic {auth}']

if not (ROOT/'.git').exists():
    subprocess.run(git+['clone',f'https://github.com/{REPO}.git',str(ROOT)],check=True)
else:
    subprocess.run(git+['-C',str(ROOT),'fetch','origin','main'],check=True)
    subprocess.run(['git','-C',str(ROOT),'checkout','main'],check=True)
    subprocess.run(git+['-C',str(ROOT),'pull','--ff-only','origin','main'],check=True)

if str(ROOT) not in sys.path: sys.path.insert(0,str(ROOT))
print('Main branch ready:',ROOT)
```

## 1. Install dependencies


```python
import subprocess, sys
subprocess.check_call([sys.executable,'-m','pip','install','-q','-r',str(ROOT/'requirements-colab.txt')])
```

## 2. Train frozen teachers

Runs the known-good frozen Wav2Vec2 and MiniLM baseline. Existing valid feature caches in Google Drive are reused. Audio cache validity follows waveform + transcript timestamps; text cache validity follows the transcript only. The script is syntax-checked before execution.


```python
import runpy, subprocess, sys
script=ROOT/'scripts/teachers/train_frozen_teachers.py'
assert script.is_file(), f'Missing script: {script}'
subprocess.run([sys.executable,'-m','py_compile',str(script)],check=True)
print('Frozen teacher script syntax verified:',script)
_=runpy.run_path(str(script),run_name='__main__')
```
